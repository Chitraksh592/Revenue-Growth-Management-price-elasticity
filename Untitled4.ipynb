```python
import pandas as pd
path = '/content/drive/MyDrive/colab/'
sales = pd.read_csv(path + "sales_train_evaluation.csv")
prices = pd.read_csv(path + "sell_prices.csv")
cal = pd.read_csv(path + "calendar.csv")
```

```python
import numpy as np

STORE = "CA_1"
CAT = "FOODS"

# 1. Taking one store and one category
sales_f = sales[(sales["store_id"] == STORE) & (sales["cat_id"] == CAT)].copy()
prices_f = prices[prices["store_id"] == STORE]

# 2. Keeping the top 100 items by the total units sold
day_cols = [c for c in sales_f.columns if c.startswith("d_")]
sales_f["total"] = sales_f[day_cols].sum(axis=1)
sales_f = sales_f.nlargest(100, "total")

# 3. Wide -> long (one row per item per day)
long = sales_f.melt(id_vars=["item_id"], value_vars=day_cols,
                    var_name="d", value_name="units")

# 4. Attaching week id and holiday info from the calendar
long = long.merge(cal[["d", "wm_yr_wk", "date", "event_name_1", "snap_CA"]], on="d")

# 5. Weekly totals per item
weekly = (long.groupby(["item_id", "wm_yr_wk"])
              .agg(units=("units", "sum"),
                   date=("date", "min"),
                   snap_days=("snap_CA", "sum"),
                   has_event=("event_name_1", lambda s: s.notna().any()))
              .reset_index())

# 6. Attached weekly price
weekly = weekly.merge(prices_f[["item_id", "wm_yr_wk", "sell_price"]],
                      on=["item_id", "wm_yr_wk"], how="left")

# Weeks where the item wasn't on sale have no price so we drop them.
weekly = weekly.dropna(subset=["sell_price"])
weekly = weekly[weekly["units"] > 0]

weekly["date"] = pd.to_datetime(weekly["date"])
weekly["log_units"] = np.log(weekly["units"])
weekly["log_price"] = np.log(weekly["sell_price"])
weekly["rel_price"] = weekly["sell_price"] / weekly.groupby("item_id")["sell_price"].transform("median")

print(weekly.shape)
print(weekly.head())
print(weekly["rel_price"].describe())
```

```python
weekly["price_changed"] = (weekly["rel_price"] - 1).abs() > 0.02

var = (weekly.groupby("item_id")
             .agg(weeks=("units", "size"),
                  pct_changed=("price_changed", "mean"),
                  min_rel=("rel_price", "min"),
                  max_rel=("rel_price", "max")))

print((var["pct_changed"] > 0.10).sum(), "items with price changes in >10% of weeks")
print(var.sort_values("pct_changed", ascending=False).head(15))
```

```python
good_items = var[var["pct_changed"] > 0.10].index
weekly_f = weekly[weekly["item_id"].isin(good_items)].copy()
print(weekly_f.shape, weekly_f["item_id"].nunique(), "items")
```

```python
import statsmodels.formula.api as smf
from sklearn.metrics import r2_score, mean_absolute_percentage_error

w = weekly_f
w["month"] = w["date"].dt.month
w["weekofyear"] = w["date"].dt.isocalendar().week.astype(int)
w["has_event"] = w["has_event"].astype(int)
w["item_code"] = w["item_id"].astype("category").cat.codes

cutoff = w["date"].max() - pd.Timedelta(weeks=52)
train = w[w["date"] <= cutoff]
test = w[w["date"] > cutoff]

ols = smf.ols("log_units ~ log_price + C(item_id) + C(month) + snap_days + has_event",
              data=train).fit()
print("Price elasticity:", round(ols.params["log_price"], 2))
print("p-value:", ols.pvalues["log_price"])

pred_ols = np.exp(ols.predict(test))
print("OLS  R2:", r2_score(test["units"], pred_ols),
      "MAPE:", mean_absolute_percentage_error(test["units"], pred_ols))
```

```python
from sklearn.ensemble import HistGradientBoostingRegressor

features = ["sell_price", "rel_price", "snap_days", "has_event",
            "month", "weekofyear", "item_code"]

gbm = HistGradientBoostingRegressor(
    categorical_features=[features.index("item_code")],
    max_iter=300, learning_rate=0.05, random_state=42)
gbm.fit(train[features], train["log_units"])

pred_gbm = np.exp(gbm.predict(test[features]))
print("GBM  R2:", r2_score(test["units"], pred_gbm),
      "MAPE:", mean_absolute_percentage_error(test["units"], pred_gbm))
```

```python
cost_share = 0.60

top_items = (w.groupby("item_id")["units"].sum().nlargest(3).index)

all_results = []
for item in top_items:
    hist = w[w["item_id"] == item].sort_values("date")
    base = hist.tail(8).copy()
    med = hist["sell_price"].median()
    cost = cost_share * med
    lo, hi = hist["rel_price"].min(), hist["rel_price"].max()

    for m in np.round(np.arange(0.70, 1.31, 0.05), 2):
        if m < lo - 1e-9 or m > hi + 1e-9:
            continue
        sim = base.copy()
        price = med * m
        sim["sell_price"] = price
        sim["rel_price"] = m
        sim["log_price"] = np.log(price)

        u_gbm = np.exp(gbm.predict(sim[features])).sum()
        u_ols = np.exp(ols.predict(sim)).sum()
        all_results.append({
            "item": item, "price_mult": m, "price": round(price, 2),
            "units_gbm": round(u_gbm), "profit_gbm": round(u_gbm * (price - cost), 1),
            "units_ols": round(u_ols), "profit_ols": round(u_ols * (price - cost), 1)})

res = pd.DataFrame(all_results)
for item in top_items:
    print(item)
    print(res[res["item"] == item].drop(columns="item").to_string(index=False))
    print()
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(1, 3, figsize=(18, 5))

# 1. Price vs units (log scale)
ax[0].scatter(w["log_price"], w["log_units"], alpha=0.1, s=5)
ax[0].set_xlabel("log price"); ax[0].set_ylabel("log units"); ax[0].set_title("Price vs demand")

# 2. Actual vs predicted (GBM, test set)
ax[1].scatter(test["units"], pred_gbm, alpha=0.2, s=5)
mx = test["units"].max()
ax[1].plot([0, mx], [0, mx], "r--")
ax[1].set_xlabel("Actual units"); ax[1].set_ylabel("Predicted units"); ax[1].set_title("GBM: actual vs predicted")

# 3. Profit vs price for top item
item = top_items[0]
r = res[res["item"] == item]
ax[2].plot(r["price_mult"], r["profit_gbm"], label="GBM")
ax[2].plot(r["price_mult"], r["profit_ols"], label="OLS")
ax[2].set_xlabel("Price (x normal)"); ax[2].set_ylabel("Profit"); ax[2].set_title(f"Profit vs price: {item}")
ax[2].legend()

plt.tight_layout()
plt.savefig("results.png", dpi=150)
plt.show()
```
